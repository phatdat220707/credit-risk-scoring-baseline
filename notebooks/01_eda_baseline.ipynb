# Credit Risk Scoring Baseline

Goal: load the UCI German Credit dataset, inspect the data, and prepare for a baseline ML model.

```python
import pandas as pd
import numpy as np

from ucimlrepo import fetch_ucirepo
```

## Load Dataset

```python
german_credit = fetch_ucirepo(id=144)

X = german_credit.data.features
y = german_credit.data.targets

type(X), type(y), X.shape, y.shape
```

```python
X.head()
```

```python
y.head()
```

## Combine Features And Target

```python
target_col = y.columns[0]

df = X.copy()
df["target"] = y[target_col]

df.head()
```
